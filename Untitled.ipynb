```python
import sqlite3

conn = sqlite3.connect('ecommerce.db')

df[['event_time', 'event_type', 'product_id', 'brand', 'price', 'user_id']].to_sql('events', conn, if_exists='replace', index=False)

print("Данные загружены в SQLite!")
```

```python
query1 = """
select brand, sum(price) as total_revenue
from events
where event_type = 'purchase'
group by brand
order by total_revenue desc
limit 5;
"""

print("ТОП-5 БРЕНДОВ:")
display(pd.read_sql_query(query1, conn))


query2 = """
select product_id, count(*) as cart_count
from events
where event_type = 'cart'
group by product_id
order by cart_count desc
limit 1;
"""

print("САМЫЙ ПОПУЛЯРНЫЙ ТОВАР В КОРЗИНЕ:")
display(pd.read_sql_query(query2, conn))
```

```python
import pandas as pd

df['event_time'] = pd.to_datetime(df['event_time'])

df['hour'] = df['event_time'].dt.hour

purchases_by_hour = df[df['event_type'] == 'purchase'].groupby('hour').size()

plt.figure(figsize=(10, 5))
plt.plot(purchases_by_hour.index, purchases_by_hour.values, marker='o', color='green')
plt.title('Количество покупок в зависимости от времени суток')
plt.xlabel('Час (0-23)')
plt.ylabel('Количество покупок')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
cart_brands = df[df['event_type'] == 'cart'].groupby('brand').size()
purchase_brands = df[df['event_type'] == 'purchase'].groupby('brand').size()

conversion = (purchase_brands / cart_brands * 100).round(2)
conversion = conversion.dropna().sort_values(ascending=False)

display(conversion.head(5))
```

```python
vip_users = df[df['event_type'] == 'purchase'].groupby('user_id')['price'].sum().reset_index()

vip_users = vip_users.sort_values(by='price', ascending=False).head(10)

vip_users
```

```python
views = df[df['event_type'] == 'view']

top_categories = views.groupby('category_id')['user_id'].nunique().reset_index(name='unique_users')

top_categories = top_categories.sort_values(by='unique_users', ascending=False).head(5)

display(top_categories)
```

```python
import pandas as pd

df = pd.read_csv('clean_ecommerce.csv')

df = df.dropna(subset=['price'])

df['price'] = df['price'].astype(str).str.replace(',', '.')
df['price'] = pd.to_numeric(df['price'], errors='coerce')

df = df.dropna(subset=['price'])

df[['event_time', 'event_type', 'brand', 'price', 'user_id']].to_csv('ecommerce_clean_final.csv', index=False)
print("Файл очищен! Осталось строк:", len(df))
```

```python
import plotly.express as px

funnel_df = df[df['event_type'] != 'remove_from_cart']

funnel_counts = funnel_df['event_type'].value_counts().reset_index()
funnel_counts.columns = ['event_type', 'count']

fig = px.funnel(funnel_counts, x='count', y='event_type', 
                title='Воронка продаж интернет-магазина косметики',
                color='event_type')
fig.show()
```

```python
import matplotlib.pyplot as plt
import plotly.express as px
import pandas as pd

df['event_time'] = pd.to_datetime(df['event_time'])
df['hour'] = df['event_time'].dt.hour
purchases_by_hour = df[df['event_type'] == 'purchase'].groupby('hour').size()

plt.figure(figsize=(10, 5))
plt.plot(purchases_by_hour.index, purchases_by_hour.values, marker='o', color='green')
plt.title('Количество покупок в зависимости от времени суток')
plt.xlabel('Час (0-23)')
plt.ylabel('Количество покупок')
plt.grid(True, alpha=0.3)
plt.savefig('purchases_by_hour.png', bbox_inches='tight')
plt.show()

funnel_df = df[df['event_type'] != 'remove_from_cart']
funnel_counts = funnel_df['event_type'].value_counts().reset_index()
funnel_counts.columns = ['event_type', 'count']
fig = px.funnel(funnel_counts, x='count', y='event_type', 
                title='Воронка продаж интернет-магазина косметики',
                color='event_type')
fig.write_image("funnel.png")

top_brands_plotly = df[df['event_type'] == 'purchase'].groupby('brand')['price'].sum().reset_index()
top_brands_plotly = top_brands_plotly.sort_values(by='price', ascending=False).head(10)
fig2 = px.bar(top_brands_plotly, x='price', y='brand', orientation='h', 
              title='Топ-10 брендов по выручке',
              labels={'price': 'Выручка (руб.)', 'brand': 'Бренд'},
              color='price')
fig2.update_layout(yaxis={'categoryorder':'total ascending'})
fig2.write_image("top_brands.png")

print("Все три графика сохранены в папку проекта!")
```
